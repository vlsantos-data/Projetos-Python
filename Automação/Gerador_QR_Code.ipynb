# Gerador de QR Code com Estilo
Este notebook contém o código para gerar QR Codes personalizados usando as bibliotecas `qrcode` e `Pillow`.

```python
import qrcode 
from qrcode.image.styledpil import StyledPilImage
from qrcode.constants import ERROR_CORRECT_H
```

```python
def criar_qrcode(link, nome_arquivo):
    qr = qrcode.QRCode(error_correction=ERROR_CORRECT_H)
    qr.add_data(link)

    imagem = qr.make_image(
        image_factory=StyledPilImage,
        # Se quiser adicionar um logo, coloque o caminho em embeded_image_path
        embeded_image_path=""
    )
    imagem.save(nome_arquivo)
    
    return imagem
```

```python
link = input("Qual é o Link: ")
nome_arquivo = input("Nome do arquivo final (ex: meu_qr.png): ")

img = criar_qrcode(link, nome_arquivo)
img # Exibe o QR Code diretamente no notebook
```
